## Task 4 – Pandas Analysis
* Group by species and calculate the average of each measurement.
* Find the species with the largest average petal length.
* Sort the dataset by sepal width in descending order.
* Add a new column with a computed feature (e.g., petal area = petal_length × petal_width

```python
iris_grouped = iris.groupby('Species').mean()
iris_grouped.head()
```

```python
print(f"The species with the largest average petal length is {iris_grouped['petal_length'].idxmax()}")
```

```python
iris.sort_values(by='sepal_width', ascending=False, inplace=True)
iris.head()
```

```python
iris["petal_area"] = iris["petal_length"]*iris["petal_width"]
iris.head()
```

## Task 5 – Visualization with Matplotlib
* Create a histogram for sepal length.
* Draw a boxplot comparing petal length across species.
* Scatter plot: petal length vs petal width, colored by species.
* Line plot: mean sepal length per species.
* Add labels, legends, and titles to make each plot presentation-ready.

```python
plt.hist(iris["sepal_length"], bins =10)
plt.title("Histogram for Sepal_length")
plt.ylabel("Count")
plt.xlabel("Sepal Length")
```

```python
plt.boxplot(iris["petal_length"])
plt.ylabel("Petal Length(cm)")
```

```python

species_map = {species: idx for idx, species in enumerate(iris['Species'].unique())}
species_codes = iris['Species'].map(species_map)
scatter = plt.scatter(iris["petal_length"], iris["petal_width"], c=species_codes)
plt.legend(handles=scatter.legend_elements()[0], labels=species_map.keys(), title="species")
plt.title("Petal Length VS Petal Width")
plt.xlabel("Petal Length(cm)")
plt.ylabel("Petal Width(cm)")
```
